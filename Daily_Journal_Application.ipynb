```python

def write_journal():
    date = input("Enter date: ")
    activity = input("Today's activity: ")
    learning = input("What did you learn: ")
    mood = input("Your mood: ")
    goal = input("Tomorrow's goal: ")

    file = open("journal.txt", "a")
    file.write("Date: " + date + "\n")
    file.write("Activity: " + activity + "\n")
    file.write("Learning: " + learning + "\n")
    file.write("Mood: " + mood + "\n")
    file.write("Tomorrow's Goal: " + goal + "\n")
    file.write("----------------------\n")
    file.close()

    print("Journal saved successfully!")


def read_journal():
    try:
        file = open("journal.txt", "r")
        print(file.read())
        file.close()
    except FileNotFoundError:
        print("No journal found")


while True:
    print("\n1. Write Journal")
    print("2. Read Journal")
    print("3. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        write_journal()
    elif choice == "2":
        read_journal()
    elif choice == "3":
        print("Thank you!")
        break
    else:
        print("Invalid choice")
```
